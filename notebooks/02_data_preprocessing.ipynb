# 2. Data Cleaning and Preparation

## 2.0 Setup and Data Loading

```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings("ignore")
```

```python
neiss_2024 = pd.read_excel('../raw_data/neiss_2024_raw.xlsx')

print("Shape:", neiss_2024.shape)
neiss_2024.info()
```

## 2.1 Column Name Standardization (Lowercase)

```python
neiss_2024_clean = neiss_2024.copy()
neiss_2024_clean.columns = neiss_2024.columns.str.lower()
neiss_2024_clean.head(2).T
```

## 2.2 Missing Value and Code Value Handling (0 = Not Recorded)

```python
missing_zero_cols = ['sex', 'race', 'hispanic', 'location',
'product_1', 'product_2', 'product_3']

for col in missing_zero_cols:
    neiss_2024_clean[col] = (neiss_2024_clean[col].replace(0, pd.NA).astype('category'))
```

## 2.3 Categorical Mapping

```python
def categorical_mapping(df, col, mapping):
  df[col] = df[col].map(mapping)
  df[col] = df[col].str.lower().astype('category')
  return df[col].unique()
```

### 2.3.1 Mapping race of the patient as recorded in the ED

```python
race_map = {1: 'white', 2: 'black/african american',
            3: 'other', 4: 'asian', 5: 'american indian/alaska native',
            6: 'native hawaiian/pacific islander'}

categorical_mapping(neiss_2024_clean, 'race', race_map)
```

### 2.3.2 Mapping self-reported sex of the patient

```python
gender_map = {1: 'male', 2: 'female'}

categorical_mapping(neiss_2024_clean, 'sex', gender_map)
```

### 2.3.3 Mapping whether the patient is Hispanic/Latino

```python
hispanic_map = {1: 'yes', 2: 'no'}

categorical_mapping(neiss_2024_clean, 'hispanic', hispanic_map)
```

### 2.3.4 Mapping primary injured body part

```python
neiss_2024_clean['body_part'] = (neiss_2024_clean['body_part'].replace({87: np.nan}))

body_part_map = { 0 : 'Internal', 30: 'Shoulder', 31 : 'Upper Trunk',
                  32 : 'Elbow', 33 : 'Lower Arm', 34 : 'Wrist',
                  35 : 'Knee', 36 : 'Lower Leg', 37 : 'Ankle',
                  38 : 'Pubic Region', 75 : 'Head',76 : 'Face',
                  77 : 'Eyeball', 79 : 'Lower Trunk', 80 : 'Upper Arm',
                  81 : 'Upper Leg', 82 : 'Hand', 83 : 'Foot',
                  84 : '25-50% of Body', 85 : 'All Parts Body', 88 : 'Mouth',
                  89 : 'Neck', 92 : 'Finger', 93 : 'Toe', 94 : 'Ear'}


categorical_mapping(neiss_2024_clean, 'body_part', body_part_map)
```

### 2.3.5 Mapping secondary injured body part (if applicable)

```python
categorical_mapping(neiss_2024_clean, 'body_part_2', body_part_map)
```

### 2.3.6 Mapping primary injury diagnosis

```python
diagnosis_map = { 41: 'Ingestion', 42: 'Aspiration', 46: 'Burns, Electrical',
                  47: 'Burns, Not Specified', 48: 'Burns, Scald', 49: 'Burns, Chemical',
                  50: 'Amputaion', 51: 'Burns, Thermal', 52: 'Concussions',
                  53: 'Contusions, Abrasions', 54: 'Crushing', 55: 'Dislocation',
                  56: 'Foreign Body', 57: 'Fracture', 58: 'Hematoma', 59: 'Laceration',
                  60: 'Dental Injury' ,61: 'Nerve Damage' ,62: 'Internal Organ Injury',
                  63: 'Puncture' ,64: 'Strain, Sprain' ,65: 'Anoxia' ,66: 'Hemorrhage',
                  67: 'Electric Shock' ,68: 'Poisoning' ,69: 'Submersion',
                  71: 'Other/Not Stated' ,72: 'Avulsion' ,73: 'Burns, Radiation' ,74: 'Dermatitis, Conjunctivitis'}

categorical_mapping(neiss_2024_clean, 'diagnosis', diagnosis_map)
```

### 2.3.7 Mapping secondary injury diagnosis (if applicable)

```python
categorical_mapping(neiss_2024_clean, 'diagnosis_2', diagnosis_map)
```

### 2.3.8 Mapping patient outcome after ED visit

```python
neiss_2024_clean['disposition'] = (neiss_2024_clean['disposition'].replace({9: np.nan}))

disposition_map = { 1: 'Treated/Examined and Released',
                   2: 'Treated and Transferred',
                   4: 'Treated and Admitted/Hospitalized',
                   5: 'Held for Observation',
                   6: 'Left Without Being Seen',
                   8: 'Fatality, Incl. DOA, Died in ER'}

categorical_mapping(neiss_2024_clean, 'disposition', disposition_map)
```

### 2.3.9 Mapping location where the injury occurred

```python
location_map = {1: 'Home', 2: 'Farm/Ranch', 4: 'Street or Highway',
                5: 'Other Public Property', 6: 'Mobile/Manufactured Home',
                7: 'Industrial', 8: 'School/Daycare', 9: 'Place of Recreation or Sports'}

categorical_mapping(neiss_2024_clean, 'location', location_map)
```

### 2.3.10 Mapping whether fire was involved in the incident

```python
fire_involvement_map = {
    0:'no_fire_or_not_recorded',
    1:'fire_fd_attended',
    2:'fire_fd_not_attended',
    3:'fire_fd_unknown'
}

categorical_mapping(neiss_2024_clean, 'fire_involvement', fire_involvement_map)
```

### 2.3.11 Mapping alcohol consumption related to the incident

```python
alcohol_map = {0: 'No/No Information', 1: 'Yes'}

categorical_mapping(neiss_2024_clean, 'alcohol', alcohol_map)
```

### 2.3.12 Mapping drug or medication involvement in the incident

```python
drug_map = {0: 'No/No Information', 1: 'Yes'}

categorical_mapping(neiss_2024_clean, 'drug', drug_map)
```

## 2.4 Transforming To Age By Year

```python
neiss_2024_clean['age_year'] = (neiss_2024_clean['age'].
                                replace(0, pd.NA).
                                astype('Float64'))

months_age = neiss_2024_clean['age_year'].between(201, 223)

neiss_2024_clean.loc[months_age, "age_year"] = (neiss_2024_clean.loc[months_age, "age_year"] - 200) / 12

neiss_2024_clean['age_year'].value_counts(dropna=False, ascending=False )
```

### 2.5.4 Handling missing data
Dropping rows with <1% missingness and encoding categorical missing values as explicit `unknown`.

```python
neiss_2024_clean = neiss_2024_clean.dropna(subset=['sex', 'age_year', 'disposition'])

categorical_fill_cols = ['location', 'race', 'hispanic', 'body_part']

for col in categorical_fill_cols:
    neiss_2024_clean[col] = (
        neiss_2024_clean[col].astype('category')
        .cat.add_categories('unknown').fillna('unknown'))
```

## 2.6 Final Analytical Dataset

- The cleaned dataset served as the final analytical dataset for all subsequent exploratory, statistical, and predictive analyses.

- **Exclusion of high-missing variables:**
  Variables with more than **40% missing values** were excluded from the dataset due to their limited analytical value and the risk of introducing bias through excessive imputation.

- **Exclusion of certain variables for xploratory Data Analysis:**
  
  - The `product_1` column was excluded from the EDA due to its high categorical complexity and the extensive recoding required to ensure stability during modeling. However, it was retained for machine learning target encoding.
  
  - The `narrative_1` column was excluded from the EDA because it contains unstructured free-text data, which requires advanced natural language processing techniques. It was, however, retained for use in machine learning target encoding.

- Survey design variables (weight, stratum, and PSU) were retained to preserve the possibility for population-level inference; however, they were not incorporated into exploratory visualizations or predictive modeling procedures.

```python
analysis_cols = [
    'cpsc_case_number',
    'treatment_date',
    'age_year',
    'sex',
    'race',
    'hispanic',
    'body_part',
    'diagnosis',
    'location',
    'fire_involvement',
    'alcohol',
    'drug',
    'disposition',
    'product_1',
    'narrative_1',
    'stratum',
    'psu',
    'weight']

neiss_2024_analysis = neiss_2024_clean[analysis_cols].copy()
neiss_2024_analysis.head(2).T
```

## 2.7 Basic Feature Engineering

### 2.7.1 Datetime Feature Creation and Summarization

```python
neiss_2024_analysis['treatment_date'] = pd.to_datetime(
    neiss_2024_analysis['treatment_date'], errors='coerce')

neiss_2024_analysis['year'] = neiss_2024_analysis['treatment_date'].dt.year
neiss_2024_analysis['month'] = neiss_2024_analysis['treatment_date'].dt.month
neiss_2024_analysis['weekday'] = neiss_2024_analysis['treatment_date'].dt.day_name()

print("Treatment Date Summary:")
print('Earliest date:', neiss_2024_analysis['treatment_date'].min())
print('Latest date:', neiss_2024_analysis['treatment_date'].max())
print('Missing values:', neiss_2024_analysis['treatment_date'].isna().sum())
```

### 2.7.2 Age Group Categorization
We categorized age into clinically relevant groups for descriptive analysis.

```python
neiss_2024_analysis['age_group'] = pd.cut(
    neiss_2024_analysis['age_year'],
    bins=[0, 1, 4, 12, 18, 35, 50, 65, 120],
    labels=[
        'infant (<1)',
        'toddler (1-4)',
        'child (5-12)',
        'adolescent (13-17)',
        'young adult (18-34)',
        'mid adult (35-49)',
        '50-64',
        'elderly (65+)'])
```

### 2.7.3 Hospital Admission as a Proxy for Injury Severity
Admission was defined as cases requiring continued inpatient care, including hospitalization or transfer. Fatal cases were excluded due to their low frequency and distinct clinical pathway.

The binary outcome variable `admitted` was coded as 1 for cases requiring hospitalization or transfer and 0 otherwise.

```python
admitted_conditions = ['treated and admitted/hospitalized',
                       'treated and transferred']

neiss_2024_analysis['admitted'] = (
    neiss_2024_analysis['disposition'].isin(admitted_conditions).astype(int))

admission_rate = neiss_2024_analysis['admitted'].mean()
print(f"Overall, {admission_rate:.2%} of cases resulted in hospital admission")
neiss_2024_analysis.head(2).T
```
