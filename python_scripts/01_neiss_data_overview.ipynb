## 1.0 Setup and Data Loading

```python
import pandas as pd
import warnings
warnings.filterwarnings("ignore")
```

```python
neiss_2024 = pd.read_excel('../raw_data/neiss_2024_raw.xlsx')

print("Shape:", neiss_2024.shape)
display(neiss_2024.head())
```

## 1.2 Checking for Detailed Information About the Data

```python
neiss_2024.info()
```

## 1.3 Evaluating Missing Data Patterns

```python
missing_summary = (
    neiss_2024.isna()
    .sum()
    .to_frame("missing_count")
)

missing_summary["missing_pct"] = (
    missing_summary["missing_count"] / len(neiss_2024) * 100
)

missing_summary.sort_values("missing_pct", ascending=False)
```

## 1.5 Distribution of Categorical Variables

```python
categorical_cols = neiss_2024.select_dtypes(include="object").columns

for col in categorical_cols:
    print(f"\n===== {col} =====")
    display(neiss_2024[col].value_counts().head(10))
```
